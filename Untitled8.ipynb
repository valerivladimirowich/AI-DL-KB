```python
# 1. Install necessary libraries (Run this if you haven't)
# !pip install ultralytics easyocr

import os
import re
import cv2
import numpy as np
import torch
from ultralytics import YOLO
import easyocr
from google.colab.patches import cv2_imshow
from google.colab import files

# -------- Helper Functions --------

def clean_text(text):
    """Removes special characters and keeps only alphanumeric capital letters."""
    text = text.upper().strip()
    return re.sub(r"[^A-Z0-9]", "", text)

def preprocess_plate(img):
    """Applies filters to make the text clearer for OCR."""
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    # Reduces noise while keeping edges sharp
    gray = cv2.bilateralFilter(gray, 11, 17, 17)
    # Binary thresholding
    thresh = cv2.threshold(
        gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )[1]
    return thresh

def recognize_plate(reader, plate_img):
    """Processes the cropped plate and performs OCR."""
    if plate_img is None or plate_img.size == 0:
        return "", 0.0

    processed = preprocess_plate(plate_img)

    results = reader.readtext(
        processed,
        detail=1,
        paragraph=False,
        allowlist="ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    )

    best_text = ""
    best_conf = 0.0

    for r in results:
        text = clean_text(r[1])
        conf = float(r[2])

        if len(text) >= 4 and conf > best_conf:
            best_text = text
            best_conf = conf

    return best_text, best_conf

def add_plate_preview_to_bottom(image, plate_crop, plate_text=""):
    """Creates a white canvas at the bottom to show the zoomed-in plate."""
    if plate_crop is None or plate_crop.size == 0:
        return image

    h, w = image.shape[:2]
    preview_w = min(320, max(180, w // 3))
    preview_h = 110

    plate_resized = cv2.resize(plate_crop, (preview_w, preview_h))

    # Create a white footer area (160 pixels high)
    canvas = np.ones((h + 160, w, 3), dtype=np.uint8) * 255
    canvas[:h, :w] = image

    x1, y1 = 20, h + 30
    x2, y2 = x1 + preview_w, y1 + preview_h

    # Draw the zoomed plate image
    cv2.rectangle(canvas, (x1 - 2, y1 - 2), (x2 + 2, y2 + 2), (0, 0, 0), 2)
    canvas[y1:y2, x1:x2] = plate_resized

    label = f"Zoomed plate: {plate_text}" if plate_text else "Zoomed plate"
    cv2.putText(
        canvas, label, (x1, y1 - 8),
        cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 0, 0), 2, cv2.LINE_AA
    )
    return canvas

# -------- Main Logic --------

# Initialize Models
reader = easyocr.Reader(['en'])
# Use a license plate specific model if you have one, otherwise yolov8n detects cars
model = YOLO('yolov8n.pt')

def run_anpr():
    print("Upload a car image from your computer...")
    uploaded = files.upload()

    for filename in uploaded.keys():
        img = cv2.imread(filename)
        if img is None:
            continue

        results = model(img)

        for r in results:
            for box in r.boxes:
                # Get coordinates
                x1, y1, x2, y2 = map(int, box.xyxy[0])

                # CROP only the detected area (The license plate)
                plate_crop = img[y1:y2, x1:x2]

                # Perform OCR on the crop
                text, conf = recognize_plate(reader, plate_crop)

                # Visualization
                final_img = add_plate_preview_to_bottom(img, plate_crop, text)

                # Draw a green box around the plate on the original image
                cv2.rectangle(final_img, (x1, y1), (x2, y2), (0, 255, 0), 3)

                cv2_imshow(final_img)
                print(f"Detected Text: {text} ({conf*100:.1f}%)")

# Execute
run_anpr()
```
